```python
df["Goals"] = df["Goals"].fillna(0)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
df[df.isnull().any(axis=1)]

df.isnull().sum().sum()
```

```python
import os
print(os.getcwd())
```

```python
# Fill missing values in statistics columns with 0
stats_columns = ["MinutesPlayed", "90s", "Goals", "Assist", "PenaltyKick", "CrdY", "CrdR"]
for col in stats_columns:
    df[col] = df[col].fillna(0)
```

```python
# Fill missing Age values with average age
df["Age"] = df["Age"].fillna(df["Age"].mean().astype(int))
```

```python
# Fill missing Nation values with 'Unknown'
df["Nation"] = df["Nation"].fillna("Unknown")
```

```python
# Check missing values again
df.isnull().sum().sum()
```

```python
# Display top 10 goal scorers
top_scorers = df.sort_values(by="Goals", ascending=False)
top_scorers[["Player", "Goals"]].head(10)
```

```python
# Top 10 assist providers
top_assists = df.sort_values(by="Assist", ascending=False)
top_assists[["Player", "Assist"]].head(10)
```

```python
# Clubs with highest total goals
club_goals = df.groupby("Team")["Goals"].sum().sort_values(ascending=False)
club_goals.head(10)
```

```python
# Select top 10 scorers
top10 = df.sort_values(by="Goals", ascending=False).head(10)

# Create figure size
plt.figure(figsize=(10,6))

# Create bar chart
plt.bar(top10["Player"], top10["Goals"])

# Add title and labels
plt.title("Top 10 Goal Scorers")
plt.xlabel("Players")
plt.ylabel("Goals")

# Rotate player names for readability
plt.xticks(rotation=20)
plt.savefig("top_scorers_chart.png")

# Display chart
plt.show()
```

```python
# Top 10 assist providers
top_assist = df.sort_values(by="Assist", ascending=False).head(10)

# Create chart
plt.figure(figsize=(10,6))

plt.bar(top_assist["Player"], top_assist["Assist"])

plt.title("Top 10 Assist Providers")
plt.xlabel("Players")
plt.ylabel("Assists")

plt.xticks(rotation=20)
plt.savefig("top_assister_chart.png")

plt.show()
```

```python
df.columns
```

```python
# Count player positions
position_counts = df["Position"].value_counts()

# Create pie chart
plt.figure(figsize=(8,8))

plt.pie(position_counts,
        labels=position_counts.index,
        autopct='%1.1f%%')

# Add title
plt.title("Player Positrion Distribution")
plt.savefig("player_position_piechart.png")

# Display chart
plt.show()
```

```python
# Create histogram for player ages

plt.figure(figsize=(10,6))

plt.hist(df["Age"], bins=10)

# Add labels and title
plt.title("Age Distribution of Players")
plt.xlabel("Age")
plt.ylabel("Number of Players")
plt.savefig("Age_didtribution_histogram.png")

# Show chart
plt.show()
```

```python
# Convert Age column to integer
df["Age"] = df["Age"].astype(int)
```

```python
df["Age"].head()
```

```python
# Select top 20 scorers
top_players = df.sort_values(by="Goals", ascending=False).head(20)

# Create figure
plt.figure(figsize=(12,8))

# Scatter plot
plt.scatter(top_players["Goals"], top_players["Assist"])

# Add player names
for i in range(len(top_players)):
    plt.text(top_players["Goals"].iloc[i],
             top_players["Assist"].iloc[i],
             top_players["Player"].iloc[i],
             fontsize=9)

# Labels and title
plt.title("Top Players: Goals vs Assists")
plt.xlabel("Goals")
plt.ylabel("Assists")

plt.grid(True)

plt.show()
```
